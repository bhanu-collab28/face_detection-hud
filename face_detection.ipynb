```python
import cv2
from ultralytics import YOLO

model = YOLO("yolov8n.pt")
face_model = YOLO("yolov8m-face.pt")

photo = cv2.imread("demo4.jpeg")
person_result = model(photo)
result = face_model(photo)

cat_person = person_result[0].plot()
cat_person_face = result[0].plot(img=cat_person)
line_width=1
font_size=0.5

cv2.imshow("my window", cat_person_face)
cv2.moveWindow("my window",0,0)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```
